```python
import pandas as pd
import matplotlib.pyplot as plt
import matplotlib.dates as mdates
```

```python
df = pd.read_csv("../data/processed/caiso_lmps.csv")
```

```python
difs = df["RT5M_AVG_LMP"] - df["DAM_LMP"]


difs.hist()
```

```python
fig, ax = plt.subplots(figsize=(12,6))

ax.plot(df["Time"], df["RT5M_AVG_LMP"] - df["DAM_LMP"], label="RT5M Avg LMP")

# Format x-axis as dates
ax.xaxis.set_major_formatter(mdates.DateFormatter("%Y-%m-%d %H:%M"))

ax.xaxis.set_major_locator(mdates.DayLocator(interval=100))  # or every day

plt.xticks(rotation=45)
plt.xlabel("Time")
plt.ylabel("LMP ($/MWh)")
plt.legend()
plt.tight_layout()
plt.show()
```
